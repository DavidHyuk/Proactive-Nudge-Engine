# Data Verification Notebook

This notebook verifies the data loading and preprocessing pipeline for the Proactive-Nudge-Engine.

```python
import sys
import os

# Add src to path
sys.path.append(os.path.abspath('../src'))

from preprocess import create_dummy_data, ProactiveDataset
from transformers import AutoTokenizer
import pandas as pd
```

## 1. Inspect Raw Data

```python
raw_data = create_dummy_data(num_samples=10)
df = pd.DataFrame(raw_data)
print("Sample Data:")
display(df.head())
```

## 2. Verify Tokenization (Trigger Task)

```python
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
dataset = ProactiveDataset(raw_data, tokenizer, task='trigger')

print(f"Dataset Size: {len(dataset)}")
sample_item = dataset[0]
print("\nSample Tokenized Input (Trigger):")
print("Input IDs shape:", sample_item['input_ids'].shape)
print("Label:", sample_item['labels'])
```

## 3. Verify Tokenization (Generation Task)

```python
gen_tokenizer = AutoTokenizer.from_pretrained('google/mt5-small')
# Filter for positive examples only
pos_data = [d for d in raw_data if d['trigger_label'] == 1]
gen_dataset = ProactiveDataset(pos_data, gen_tokenizer, task='generation')

if len(gen_dataset) > 0:
    sample_gen = gen_dataset[0]
    print("\nSample Tokenized Input (Generation):")
    print("Input IDs shape:", sample_gen['input_ids'].shape)
    print("Labels shape:", sample_gen['labels'].shape)
else:
    print("No positive examples generated in random batch.")
```
